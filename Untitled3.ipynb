```python
import numpy as np

def sigmoid(x):
    return 1/(1+np.exp(-x))

def sigmoid_derivative(x):
    return x*(1-x)

def train():
    x = np.array([1,0])
    target = np.array([1,0])

    w = np.random.rand(8)
    b = np.random.rand(4)
    lr = 0.1

    for epoch in range(1000):

        # ---------------- Q1: Hidden Layer ----------------
        i1 = x[0]*w[0] + x[1]*w[2] + b[0]
        i2 = x[0]*w[1] + x[1]*w[3] + b[1]

        a1 = sigmoid(i1)
        a2 = sigmoid(i2)

        # ---------------- Q2: Output Layer ----------------
        i3 = a1*w[4] + a2*w[6] + b[2]
        i4 = a1*w[5] + a2*w[7] + b[3]

        y1 = sigmoid(i3)
        y2 = sigmoid(i4)

        # ---------------- Q3: Loss ----------------
        loss = 0.5*((target[0]-y1)**2 + (target[1]-y2)**2)

        # ---------------- Q4: Output Errors ----------------
        err1 = -(target[0]-y1)*y1*(1-y1)
        err2 = -(target[1]-y2)*y2*(1-y2)

        # ---------------- Q5: Gradients (Output Layer) ----------------
        dw5 = err1 * a1
        dw6 = err2 * a1
        dw7 = err1 * a2
        dw8 = err2 * a2

        db3 = err1
        db4 = err2

        # ---------------- Q6: Hidden Layer Errors ----------------
        err_h1 = sigmoid_derivative(a1) * (err1*w[4] + err2*w[5])
        err_h2 = sigmoid_derivative(a2) * (err1*w[6] + err2*w[7])

        # ---------------- Q7: Gradients (Hidden Layer) ----------------
        dw1 = err_h1 * x[0]
        dw2 = err_h2 * x[0]
        dw3 = err_h1 * x[1]
        dw4 = err_h2 * x[1]

        db1 = err_h1
        db2 = err_h2

        # -------- Weight Updates --------
        w[0] -= lr * dw1
        w[1] -= lr * dw2
        w[2] -= lr * dw3
        w[3] -= lr * dw4

        w[4] -= lr * dw5
        w[5] -= lr * dw6
        w[6] -= lr * dw7
        w[7] -= lr * dw8

        # -------- Bias Updates --------
        b[0] -= lr * db1
        b[1] -= lr * db2
        b[2] -= lr * db3
        b[3] -= lr * db4

        # ---------------- Q8: Epoch Output ----------------
        if epoch % 100 == 0:
            print("Epoch:", epoch, "Loss:", loss)

train()
```

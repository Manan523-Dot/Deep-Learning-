```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense

df = pd.read_excel("yahoo_data.xlsx")

# Clean column names
df.columns = df.columns.str.strip()

print("Columns:", df.columns)  # for verification

# Automatically pick a meaningful column
if 'Close' in df.columns:
    data = df['Close'].values.reshape(-1,1)
elif 'Adj Close' in df.columns:
    data = df['Adj Close'].values.reshape(-1,1)
else:
    # fallback: first numeric column
    data = df.select_dtypes(include=['float64','int64']).iloc[:,0].values.reshape(-1,1)

scaler = MinMaxScaler()
data = scaler.fit_transform(data)

def create_dataset(data, time_step=10):
    X, y = [], []
    for i in range(len(data)-time_step-1):
        X.append(data[i:(i+time_step), 0])
        y.append(data[i+time_step, 0])
    return np.array(X), np.array(y)

X, y = create_dataset(data)
X = X.reshape(X.shape[0], X.shape[1], 1)

model = Sequential([
    SimpleRNN(50, activation='relu', input_shape=(10,1)),
    Dense(1)
])

model.compile(optimizer='adam', loss='mse')
model.fit(X, y, epochs=8, validation_split=0.2)

print("Q15 (Stock) Done")
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense

df = pd.read_csv("travel_times.csv")

df.columns = df.columns.str.strip()

print("Columns:", df.columns)

# Select numeric column safely
data = df.select_dtypes(include=['int64','float64']).iloc[:,0].values.reshape(-1,1)

scaler = MinMaxScaler()
data = scaler.fit_transform(data)

def create_dataset(data, time_step=5):
    X, y = [], []
    for i in range(len(data)-time_step-1):
        X.append(data[i:(i+time_step), 0])
        y.append(data[i+time_step, 0])
    return np.array(X), np.array(y)

X, y = create_dataset(data)
X = X.reshape(X.shape[0], X.shape[1], 1)

model = Sequential([
    SimpleRNN(40, activation='relu', input_shape=(5,1)),
    Dense(1)
])

model.compile(optimizer='adam', loss='mse')
model.fit(X, y, epochs=8, validation_split=0.2)

print("Q15 (Travel) Done")
```

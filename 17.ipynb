```python
import pandas as pd
import nltk
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer

# Download tokenizer (only first time)
nltk.download('punkt')

# Load dataset
df = pd.read_csv("imdb_top_1000.csv")

# Clean column names
df.columns = df.columns.str.strip()

print("Columns:", df.columns)

# Select text column safely
if 'Overview' in df.columns:
    text = df['Overview'].dropna().iloc[0]
elif 'Description' in df.columns:
    text = df['Description'].dropna().iloc[0]
else:
    # fallback: pick first text column
    text = df.select_dtypes(include=['object']).iloc[:,0].dropna().iloc[0]

# Tokenization
tokens = word_tokenize(text)

# Stemming
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in tokens]

# Output
print("\nOriginal Text:\n", text)
print("\nFirst 20 Tokens:\n", tokens[:20])
print("\nFirst 20 Stemmed Words:\n", stemmed_words[:20])
```

```python
import numpy as np

# OR GATE
X = np.array([[0,0], [0,1], [1,0], [1,1]])
y = np.array([0, 1, 1, 1])   

weights = np.zeros(2)
bias    = 0
lr      = 0.1           
epochs  = 10

def step(z):
    return 1 if z >= 0 else 0

# Training
for epoch in range(epochs):
    errors = 0
    for xi, yi in zip(X, y):
        z    = np.dot(weights, xi) + bias
        pred = step(z)
        err  = yi - pred            
        weights += lr * err * xi   
        bias    += lr * err
        errors  += int(err != 0)
    print(f"Epoch {epoch+1}: errors={errors}, w={weights}, b={bias:.1f}")

print("\n--- Predictions ---")
for xi, yi in zip(X, y):
    pred = step(np.dot(weights, xi) + bias)
    print(f"Input {xi} → pred={pred}, actual={yi}")

print("\nFinal weights:", weights)
```

```python
import numpy as np

# NOT GATE
X = np.array([[0], [1]])
y = np.array([1, 0])   

weights = np.zeros(1)
bias    = 0
lr      = 0.1
epochs  = 10

def step(z):
    return 1 if z >= 0 else 0


for epoch in range(epochs):
    errors = 0
    for xi, yi in zip(X, y):
        z    = np.dot(weights, xi) + bias
        pred = step(z)
        err  = yi - pred
        weights += lr * err * xi
        bias    += lr * err
        errors  += int(err != 0)
    print(f"Epoch {epoch+1}: errors={errors}, w={weights}, b={bias:.1f}")


print("\n--- Predictions ---")
for xi, yi in zip(X, y):
    pred = step(np.dot(weights, xi) + bias)
    print(f"Input {xi} → pred={pred}, actual={yi}")

print("\nFinal weights:", weights)
```
